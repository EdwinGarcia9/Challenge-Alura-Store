### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
ing1 = tienda['Precio'].sum()
ing2 = tienda2['Precio'].sum()
ing3 = tienda3['Precio'].sum()
ing4 = tienda4['Precio'].sum()
print(f"El ingreso de la tienda 1 es {ing1}")
print(f"El ingreso de la tienda 2 es {ing2}")
print(f"El ingreso de la tienda 3 es {ing3}")
print(f"El ingreso de la tienda 4 es {ing4}")
```

# 2. Ventas por categoría

```python
Cat1 = tienda.groupby('Categoría del Producto').size().reset_index(name='Cantidad').sort_values("Cantidad",ascending=False)
Cat2 = tienda2.groupby('Categoría del Producto').size().reset_index(name='Cantidad').sort_values("Cantidad",ascending=False)
Cat3 = tienda3.groupby('Categoría del Producto').size().reset_index(name='Cantidad').sort_values("Cantidad",ascending=False)
Cat4 = tienda4.groupby('Categoría del Producto').size().reset_index(name='Cantidad').sort_values("Cantidad",ascending=False)
print("=====================================================")
print(Cat1)
print("=====================================================")
print(Cat2)
print("=====================================================")
print(Cat3)
print("=====================================================")
print(Cat4)
```

# 3. Calificación promedio de la tienda


```python
Cal1 = tienda.groupby('Categoría del Producto')["Calificación"].mean().reset_index(name='Calificación').sort_values("Calificación",ascending=False)
Cal2 = tienda2.groupby('Categoría del Producto')["Calificación"].mean().reset_index(name='Calificación').sort_values("Calificación",ascending=False)
Cal3 = tienda3.groupby('Categoría del Producto')["Calificación"].mean().reset_index(name='Calificación').sort_values("Calificación",ascending=False)
Cal4 = tienda4.groupby('Categoría del Producto')["Calificación"].mean().reset_index(name='Calificación').sort_values("Calificación",ascending=False)
print(f"La calificación promedio de la tienda 1 es {Cal1}")
print(f"La calificación promedio de la tienda 2 es {Cal2}")
print(f"La calificación promedio de la tienda 3 es {Cal3}")
print(f"La calificación promedio de la tienda 4 es {Cal4}")
```

# 4. Productos más y menos vendidos

```python
Top1 = tienda.groupby('Producto').size().reset_index(name='Cantidad').sort_values("Cantidad",ascending=False)
Top2 = tienda2.groupby('Producto').size().reset_index(name='Cantidad').sort_values("Cantidad",ascending=False)
Top3 = tienda3.groupby('Producto').size().reset_index(name='Cantidad').sort_values("Cantidad",ascending=False)
Top4 = tienda4.groupby('Producto').size().reset_index(name='Cantidad').sort_values("Cantidad",ascending=False)
print("=====================================================")
print(f"el top de la tienda 1 es {Top1}")
print("=====================================================")
print(f"el top de la tienda 2 es {Top2}")
print("=====================================================")
print(f"el top de la tienda 3 es {Top3}")
print("=====================================================")
print(f"el top de la tienda 4 es {Top4}")
```

# 5. Envío promedio por tienda

```python
send1 = tienda['Costo de envío'].mean()
send2 = tienda2['Costo de envío'].mean()
send3 = tienda3['Costo de envío'].mean()
send4 = tienda4['Costo de envío'].mean()

print(f"El costo de envio promedio para la tienda 1 es {send1}")
print(f"El costo de envio promedio para la tienda 2 es {send2}")
print(f"El costo de envio promedio para la tienda 3 es {send3}")
print(f"El costo de envio promedio para la tienda 4 es {send4}")
```

# 6. Graficos

```python
import matplotlib.pyplot as plt

ingresos = [ing1, ing2, ing3, ing4]
tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]

plt.bar(tiendas, ingresos)
plt.xlabel("Tiendas")
plt.ylabel("Ingresos")
plt.title("Ingresos por Tienda")
plt.show
```

```python
cats = [Cat1, Cat2, Cat3, Cat4]
nombres = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]

for df, nombre in zip(cats, nombres):
    plt.figure(figsize=(8, 4))
    plt.bar(df["Categoría del Producto"], df["Cantidad"])
    plt.xticks(rotation=45)
    plt.title(f"Categorías más vendidas - {nombre}")
    plt.tight_layout()
    plt.show()
```

```python
envios = [send1, send2, send3, send4]
tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]

plt.bar(tiendas, envios)
plt.xlabel("Tiendas")
plt.ylabel("Costo de envío promedio")
plt.title("Costo de envío promedio por tienda")
plt.show()
```
